## Part 1: L-System Foundation

The following code generates an L-system instruction. This funcation was provided in lab 3 resources

```python
def create_l_system(iterations, axiom, rules):

    result = axiom
    for _ in range(iterations):
        new_string = ""
        for char in result:
            new_string += rules.get(char, char)
        result = new_string
    return result


```

## Part 2: Mosaic Drawing System

The generated L-System string is interpreted using Turtle graphics.

- F draws a line forward.
- f moves forward without drawing.
- '+' turns the Turtle to the right.
- '-' turns the Turtle to the left.

The drawing function also accepts color and line thickness parameters.

```python
def draw_l_system(instructions, angle, distance,
                  color="black", thickness=2):
    """
    Draw an L-System using Turtle graphics.

    Parameters:
    - instructions: generated L-System command string
    - angle: turning angle in degrees
    - distance: length of each forward movement
    - color: drawing color
    - thickness: width of the drawn lines
    """

    t.pencolor(color)
    t.pensize(thickness)

    for cmd in instructions:

        if cmd == 'F':
            # Move forward while drawing
            t.forward(distance)

        elif cmd == 'f':
            # Move forward without drawing
            t.penup()
            t.forward(distance)
            t.pendown()

        elif cmd == '+':
            # Turn right
            t.right(angle)

        elif cmd == '-':
            # Turn left
            t.left(angle)
```

```python
def setup_turtle(background="white"):
    #Prepare the canvas for drawing mosaic patterns.

    t.initializeTurtle(
        initial_window_size=(600, 300)
    )

    t.hideturtle()
    t.speed(13)
    t.bgcolor(background)


def move_turtle(x, y, heading=0):
    #Move the turtle to a tile position without drawing.

    t.penup()
    t.goto(x, y)
    t.setheading(heading)
    t.pendown()
```

## Part 3: Mosaic Pattern Rules

Four L-System pattern families are used to create different geometric
mosaic structures. Each design has its own axiom, production rule, angle,
and default visual parameters.

The four patterns are:

1. **Square Mosaic** ~ dense rectilinear geometry created using 90-degree turns.
2. **Star Mosaic** ~ repeated angular forms that create a decorative star-like structure.
3. **Cross Mosaic** ~ symmetrical geometric forms based on repeated right-angle turns.
4. **Snowflake Mosaic** ~  geometry using 60-degree turns to create a detailed snowflake-like pattern.

Although all four patterns use the same L-System engine, differences in their axioms, production rules, angles, and parameters produce distinct visual structures.

```python
# Geometric L-System rules adapted from Paul Bourke's L-System examples.
# Source: https://paulbourke.net/fractals/lsys/


mosaic_patterns = {

    "square": {
        "name": "Square Mosaic",

        "axiom": "F+F+F+F",

        "rules": {
            "F": "FF+F-F+F+FF"
        },

        "angle": 90,
        "iterations": 3,
        "distance": 5,
        "color": "navy",
        "thickness": 2,
        "background": "white",


        "positions": [
            (170, 160),
            (340, 160),
            (510, 160)
        ]
    },


"cross": {
    "name": "Geometric Cross Tile",

    "axiom": "F+F+F+F",

    "rules": {
        "F": "F+F-F-F+F"
    },

    "angle": 90,
    "iterations": 3,
    "distance": 5,
    "color": "purple",
    "thickness": 2,
    "background": "white",
    "positions": [
        (80, 100),
        (230, 100),
        (380, 100)
      ]

},


    "star": {
        "name": "Star Mosaic",

        "axiom": "F++F++F++F",

        "rules": {
            "F": "F+F--F+F"
        },

        "angle": 45,
        "iterations": 3,
        "distance": 4,
        "color": "darkred",
        "thickness": 2,
        "background": "white",
         "positions": [
            (80, 100),
            (230, 100),
            (380, 100)
        ]
    },

    #I know only three pattern shapes are required but I really wanted to try snowflake patterns too for the tiles

    "snowflake": {
    "name": "Snowflake Star Tile",

    "axiom": "F++F++F",

    "rules": {
        "F": "F-F++F-F"
    },

    "angle": 60,
    "iterations": 3,
    "distance": 4,
    "color": "teal",
    "thickness": 2,
    "background": "white",
     "positions": [
            (80, 100),
            (230, 100),
            (380, 100)
        ]
}




}
```

## Part 4: Mosaic Generator

The `generate_mosaic()` function connects the L-System rules to the drawing
system. A mosaic family is selected and its default parameters are loaded.

Individual parameters can also be overridden. This allows multiple artworks
to be generated from the same underlying L-System while experimenting with
angle, iterations, segment length, color, and line thickness.

```python
def generate_mosaic(pattern_type,
                    iterations=None,
                    angle=None,
                    distance=None,
                    color=None,
                    thickness=None,
                    background=None):

    #Generate and draw a selected L-System mosaic pattern.


    if pattern_type not in mosaic_patterns:
        print("Unknown mosaic pattern.")
        print("Available patterns:", list(mosaic_patterns.keys()))
        return

    pattern = mosaic_patterns[pattern_type]


    if iterations is None:
        iterations = pattern["iterations"]

    if angle is None:
        angle = pattern["angle"]

    if distance is None:
        distance = pattern["distance"]

    if color is None:
        color = pattern["color"]

    if thickness is None:
        thickness = pattern["thickness"]

    if background is None:
        background = pattern["background"]

    # Generate the instruction strings
    instructions = create_l_system(
        iterations,
        pattern["axiom"],
        pattern["rules"]
    )

    # Prepare the drawing canvas
    setup_turtle(background)

    # Drawing the generated mosaic
    draw_l_system(
        instructions,
        angle,
        distance,
        color,
        thickness
    )

    # Display the parameters I used
    print("Pattern:", pattern["name"])
    print("Iterations:", iterations)
    print("Angle:", angle)
    print("Segment Length:", distance)
    print("Color:", color)
    print("Line Thickness:", thickness)
    print("Background:", background)
```

```python
def generate_tile_row(pattern_type,
                      iterations=None,
                      angle=None,
                      distance=None,
                      color=None,
                      thickness=None,
                      background=None,
                      random_colors=False):

    #Generate three repeated L-System mosaic tiles in one row.

    # Check if the selected pattern exists
    if pattern_type not in mosaic_patterns:
        print("Unknown mosaic pattern.")
        print("Available patterns:", list(mosaic_patterns.keys()))
        return

    pattern = mosaic_patterns[pattern_type]

    # Use default pattern values if no custom values are provided
    if iterations is None:
        iterations = pattern["iterations"]

    if angle is None:
        angle = pattern["angle"]

    if distance is None:
        distance = pattern["distance"]

    if color is None:
        color = pattern["color"]

    if thickness is None:
        thickness = pattern["thickness"]

    if background is None:
        background = pattern["background"]

    # Generate the L-System instruction string once
    instructions = create_l_system(
        iterations,
        pattern["axiom"],
        pattern["rules"]
    )

    # Prepare the canvas
    setup_turtle(background)

    # Starting positions for three adjacent tiles
    tile_positions = pattern["positions"]

    # Available colors for randomization
    color_palette = [
        "teal",
        "purple",
        "navy",
        "darkred",
        "darkgreen",
        "orange",
        "violet",
        "pink"
    ]

    # Decide which colors the tiles will use
    if random_colors:
        # Choose 3 different random colors
        tile_colors = random.sample(color_palette, 3)
    else:
        # Use the selected color for all 3 tiles
        tile_colors = [color] * 3

    # Draw the three tiles
    for i, (x, y) in enumerate(tile_positions):

        move_turtle(x, y, 0)

        draw_l_system(
            instructions,
            angle,
            distance,
            tile_colors[i],
            thickness
        )

    # Display generation information
    print("Pattern:", pattern["name"])
    print("Tiles:", 3)
    print("Iterations:", iterations)
    print("Angle:", angle)
    print("Segment Length:", distance)
    print("Tile Colors:", tile_colors)
    print("Random Colors:", random_colors)
    print("Line Thickness:", thickness)
    print("Background:", background)
```

## Part 6: Technical Documentation

This Generative Mosaic Pattern Designer is structured around an L-System rewriting process and a Turtle-based drawing system. The `create_l_system()` function begins with an axiom and repeatedly applies a set of production rules for a specified number of iterations.

The resulting instruction string is passed to `draw_l_system()`, where `F` represents forward drawing and the `+`
and `-` symbols control right and left turns. The `generate_tile_row()`function builds on by drawing the generated pattern three times at
adjacent positions, creating the appearance of a repeating mosaic pattern.

The system contains four patterns that I wanted to try: Square, Star, Cross, and Snowflake. Each uses its own combination of axiom, production rule, angle, and visual parameters. The geometric L-System rules were informed by established L-System examples and then incorporated into my own mosaic tile generator. The user can adjust iterations, angle, segment distance, color, line
thickness, and background. I also implemented an optional `random_colors`parameter, because I think its important especially for tile design. When enabled, three different colors are randomly selected.

The L-System determines the geometry by recursively rewriting its symbols, while the Turtle interpreter converts those symbols into lines and turns. Repeating the resulting motif creates the tile arrangement, while parameters such as line thickness and color change its visual texture without necessarily changing the production rules written previously.

A major challenge was that the different L-System rules did not occupy the canvas the way I wanted it too. Some patterned tiles  expanded in different directions or became too large at higher iteration counts. I also could not figure out how to properly make more iterations of tiles without the whole Colab page freezing, so decided to just do three adjacent tiles for now.


I addressed these problems by reducing the canvas size, and assigning positions and distances to the different tile patterns so they are evenly adjacent to eachother. Another design challenge was making the outputs resemble a mosaic rather than isolated fractals. Repeating each iteration three times in a row created a clearer tile-like composition.

So for the future I definitely want to work on making the tile patterns more complex and hopefully generate grids of tile patterns rather than just three tiles adjacent together.
